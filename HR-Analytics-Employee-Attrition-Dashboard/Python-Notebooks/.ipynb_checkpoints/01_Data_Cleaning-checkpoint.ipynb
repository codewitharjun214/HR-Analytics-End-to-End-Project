# What is the Employee Attrition Rate?
What percentage of employees have left the company?

```python
df["Attrition"].value_counts()
```

```python
attrition_rate = (df["Attrition"].value_counts(normalize=True) * 100).round(2)

attrition_rate
```

# Which department has the highest employee attrition?

```python
df.groupby("Department")["Attrition"].value_counts().unstack()
```

```python
department_attrition = (
    df.groupby("Department")["Attrition"]
      .value_counts(normalize=True)
      .mul(100)
      .rename("Percentage")
      .reset_index()
)

department_attrition
```

```python
department_attrition[department_attrition["Attrition"] == "Yes"]
```

# Do employees who work overtime leave the company more often?

```python
df.groupby("OverTime")["Attrition"].value_counts().unstack()
```

```python
overtime_attrition = (
    df.groupby("OverTime")["Attrition"]
      .value_counts(normalize=True)
      .mul(100)
      .rename("Percentage")
      .reset_index()
)

overtime_attrition[overtime_attrition["Attrition"] == "Yes"]
```

# Does Job Satisfaction affect employee attrition?

```python
job_satisfaction_attrition = (
    df.groupby("JobSatisfaction")["Attrition"]
      .value_counts(normalize=True)
      .mul(100)
      .rename("Percentage")
      .reset_index()
)

job_satisfaction_attrition[job_satisfaction_attrition["Attrition"] == "Yes"]
```

```python
plot_data = job_satisfaction_attrition[
    job_satisfaction_attrition["Attrition"] == "Yes"
]

plot_data
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    plot_data["JobSatisfaction"],
    plot_data["Percentage"]
)

plt.title("Job Satisfaction vs Employee Attrition Rate")
plt.xlabel("Job Satisfaction")
plt.ylabel("Attrition Rate (%)")

plt.xticks([1, 2, 3, 4], ["Low", "Medium", "High", "Very High"])

plt.show()
```

# Which Age Group has the highest employee attrition?

```python
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[18, 25, 35, 45, 55, 60],
    labels=[
        "18-25",
        "26-35",
        "36-45",
        "46-55",
        "56+"
    ],
    include_lowest=True
)

df[["Age", "AgeGroup"]].head()
```

```python
age_group_attrition = (
    df.groupby("AgeGroup")["Attrition"]
      .value_counts(normalize=True)
      .mul(100)
      .rename("Percentage")
      .reset_index()
)

age_group_attrition
```

```python
age_group_attrition[
    age_group_attrition["Attrition"] == "Yes"
]
```

# Does Monthly Income affect Employee Attrition?

```python
df["IncomeGroup"] = pd.cut(
    df["MonthlyIncome"],
    bins=[0, 5000, 10000, 15000, 20000],
    labels=[
        "0-5K",
        "5K-10K",
        "10K-15K",
        "15K-20K"
    ],
    include_lowest=True
)
```

```python
df[["MonthlyIncome", "IncomeGroup"]].head(10)
```

```python
income_attrition = (
    df.groupby("IncomeGroup")["Attrition"]
      .value_counts(normalize=True)
      .mul(100)
      .rename("Percentage")
      .reset_index()
)

income_attrition[income_attrition["Attrition"] == "Yes"]
```

```python
df["MonthlyIncome"].min()
```

```python
df["MonthlyIncome"].max()
```

# Do new employees leave more often?

```python
df["ExperienceGroup"] = pd.cut(
    df["YearsAtCompany"],
    bins=[0, 2, 5, 10, df["YearsAtCompany"].max() + 1],
    labels=[
        "0-2 Years",
        "3-5 Years",
        "6-10 Years",
        "10+ Years"
    ],
    include_lowest=True
)
```

```python
experience_attrition = (
    df.groupby("ExperienceGroup")["Attrition"]
      .value_counts(normalize=True)
      .mul(100)
      .rename("Percentage")
      .reset_index()
)

experience_attrition[
    experience_attrition["Attrition"] == "Yes"
]
```
