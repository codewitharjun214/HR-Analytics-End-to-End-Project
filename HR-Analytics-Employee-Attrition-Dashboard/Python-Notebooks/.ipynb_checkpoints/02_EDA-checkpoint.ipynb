```python
numerical_columns = [
    "Age",
    "MonthlyIncome",
    "DistanceFromHome",
    "TotalWorkingYears",
    "YearsAtCompany",
    "PercentSalaryHike",
    "NumCompaniesWorked"
]
```

```python
for col in numerical_columns:

    print("="*70)
    print(f"Column : {col}")
    print("="*70)

    print(df[col].describe())

    plt.figure(figsize=(8,4))
    plt.hist(df[col], bins=20)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()

    plt.figure(figsize=(8,2))
    plt.boxplot(df[col], vert=False)
    plt.title(f"Boxplot of {col}")
    plt.show()
```

# Phase 2 Observations

### Age

### MonthlyIncome
- Highly right-skewed.
- Contains several high-income outliers.

### DistanceFromHome
- Employees live at varying distances.
- No strong concentration.

### TotalWorkingYears
- Positively skewed.
- Many employees have relatively fewer years of experience.

### YearsAtCompany
- Most employees have spent fewer years in the company.
- Long-tenure employees are fewer.

### PercentSalaryHike
- Salary hikes are concentrated within a narrow range.

### NumCompaniesWorked
- Most employees have worked in only a few companies.

# Categorical Analysis

```python
categorical_columns = [
    "Department",
    "JobRole",
    "Gender",
    "BusinessTravel",
    "MaritalStatus",
    "EducationField",
    "OverTime",
    "Attrition"
]
```

```python
for col in categorical_columns:

    print("=" * 70)
    print(f"Column : {col}")
    print("=" * 70)

    frequency = df[col].value_counts()

    percentage = round(
        df[col].value_counts(normalize=True) * 100,
        2
    )

    summary = pd.DataFrame({
        "Frequency": frequency,
        "Percentage": percentage
    })

    print(summary)

    plt.figure(figsize=(8,4))
    plt.bar(summary.index.astype(str), summary["Frequency"])
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.xticks(rotation=45)
    plt.show()
```

# Phase 3 Observations

### Department
- Research & Development has the largest workforce.
- Human Resources has the fewest employees.

### Job Role
- Workforce is distributed across multiple job roles.
- Some roles have significantly more employees than others.

### Gender
- Male employees slightly outnumber female employees.

### Business Travel
- Most employees travel rarely.
- A smaller proportion travels frequently.

### Marital Status
- Workforce consists of single, married, and divorced employees.

### Education Field
- Life Sciences and Medical are the dominant education backgrounds.

### OverTime
- Majority of employees do not work overtime.

### Attrition
- Most employees stay with the company.
- Around 16% of employees have left the organization.

# Correlation Analysis

```python
correlation = df.corr(numeric_only=True)

correlation
```

```python
plt.figure(figsize=(14,10))

plt.imshow(correlation, cmap="coolwarm", aspect="auto")

plt.colorbar()

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=90
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title("Correlation Heatmap")

plt.tight_layout()

plt.show()
```

```python
correlation.unstack().sort_values(ascending=False)
```

# Phase 4 Observations

- Most numerical variables have weak to moderate correlations.
- Monthly Income is positively correlated with Total Working Years.
- Years At Company is positively correlated with Total Working Years.
- Years In Current Role and Years With Current Manager show a strong positive correlation.
- No severe multicollinearity is observed among most variables.

# Outlier Analysis

```python
outlier_columns = [
    "MonthlyIncome",
    "Age",
    "DistanceFromHome",
    "TotalWorkingYears",
    "YearsAtCompany"
]

for col in outlier_columns:

    plt.figure(figsize=(8,2))
    plt.boxplot(df[col], vert=False)
    plt.title(f"Outliers in {col}")
    plt.show()
```

# Phase 5 Observations

- Monthly Income contains several high-value outliers.
- Total Working Years has a few extreme observations.
- Years At Company shows some long-tenure outliers.
- Age has only a small number of outliers.
- Distance From Home has no significant outliers.
